```python
# Dashboard Charts
def show_charts():
    if not os.path.exists("student_data.csv"):
        print("No data yet!")
        return
    data = pd.read_csv("student_data.csv")
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    fig.suptitle(" College Admission Dashboard")
    # Chart 1
    data["Course"].value_counts().plot(kind="bar", ax=axes[0], color="skyblue")
    axes[0].set_title("Students per Course")
    # Chart 2
    data["Percentage"] = pd.to_numeric(data["Percentage"], errors='coerce')
    axes[1].hist(data["Percentage"], bins=5, color="lightgreen", edgecolor="black")
    axes[1].set_title("Percentage Distribution")
    plt.tight_layout()
    plt.savefig("dashboard.png")
    plt.show()
    print("Dashboard saved!")
print("Charts ready!")
```
